```python
#importing libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#loading dataset
df=pd.read_csv('cleaned_hotel_bookings.csv')
```

```python
df.head(100)
```

# Descriptive statistics

### Numerical Satistics

```python
num_cols=df.select_dtypes(include='number').columns
df[num_cols].describe()
```

### Categorical Statistics

```python
cat_cols=df.select_dtypes(include='object').columns
df[cat_cols].describe()
```

# Univariate Analysis

```python
# 1. Hotel Type Count
plt.subplot(1, 2, 1)
sns.countplot(data=df, x='hotel',palette='pastel')
plt.title('Booking Counts by Hotel Type')
plt.xlabel('Hotel Type')
plt.ylabel('Count')
plt.show()
```

```python
# 2. Cancellation Ratio
plt.subplot(1, 2, 2)
sns.countplot(data=df, x='is_canceled',palette='muted')
plt.title('Proportion of Canceled Bookings (1 = Canceled)')
plt.xlabel('Is Canceled')
plt.ylabel('Count')

plt.tight_layout()
plt.show()
```

```python
# 3. Distribution of Lead Time
plt.figure(figsize=(8, 4))
sns.histplot(data=df, x='lead_time', bins=50, kde=True, color='purple')
plt.title('Distribution of Lead Time')
plt.xlabel('Lead Time (Days)')
plt.show()
```

```python
# 4. Distribution of length of stay
plt.figure(figsize=(8, 4))
sns.histplot(data=df[df['stay_length'] < 15], x='total_nights',bins=20, color='purple', kde=True)
plt.title('Distribution of Total Length of Stay')
plt.xlabel('Total Nights Spent')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

# Multivariate Analysis

```python
# 1. Correlation Heatmap
plt.figure(figsize=(8, 6))
numerical_cols = ['lead_time', 'adr', 'stay_length', 'is_canceled', 'is_repeated_guest', 'total_of_special_requests']
sns.heatmap(df[numerical_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Matrix')
plt.show()
```

```python
# 2. Market Segment vs. ADR (Box Plot)
plt.figure(figsize=(10, 5))
sns.boxplot(data=df[df['adr'] < 1000], x='market_segment', y='adr', palette='Set2')
plt.title('Average Daily Rate (ADR) Distribution by Market Segment')
plt.xticks(rotation=45)
plt.show()
```

```python
# 3. Lead time vs ADR
plt.figure(figsize=(8, 4))
scatter_df = df[(df['adr'] > 0) & (df['adr'] < 600)]
sample_df = scatter_df.sample(n=min(2000, len(scatter_df)), random_state=42)
sns.scatterplot(
    data=sample_df, 
    x='lead_time', 
    y='adr', 
    hue='is_canceled', 
    palette='muted', 
    alpha=0.6
)
plt.title('Relationship Between Lead Time and ADR by Booking Outcome')
plt.xlabel('Lead Time (Days)')
plt.ylabel('Average Daily Rate (ADR)')
plt.legend(title='Is Canceled (1=Yes)')
plt.tight_layout()
plt.show()
```
